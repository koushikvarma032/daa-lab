```python
import time

# Number of vertices
n = int(input("Enter number of vertices: "))

# Enter adjacency matrix
print("Enter the adjacency matrix:")
graph = []

for i in range(n):
    row = list(map(int, input().split()))
    graph.append(row)

start_time = time.time()

# Store selected vertices
selected = [False] * n
selected[0] = True

edges = 0
total_cost = 0

print("\nEdges in Minimum Spanning Tree:")

while edges < n - 1:
    minimum = float('inf')
    x = 0
    y = 0

    for i in range(n):
        if selected[i]:
            for j in range(n):
                if not selected[j] and graph[i][j] != 0:
                    if graph[i][j] < minimum:
                        minimum = graph[i][j]
                        x = i
                        y = j

    print(f"{x + 1} - {y + 1} : {minimum}")

    total_cost += minimum
    selected[y] = True
    edges += 1

end_time = time.time()

print("\nTotal cost of MST:", total_cost)
print("Execution time:", end_time - start_time, "seconds")
```
